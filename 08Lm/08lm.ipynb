```python
from google import genai

API_KEY = apy_key

client = genai.Client(api_key=API_KEY)

# Usa el nombre EXACTO como aparece en la lista: 'models/gemini-2.5-flash'
response = client.models.generate_content(
    model='models/gemini-2.5-flash',  # ← Con "models/" al inicio
    contents='Cuando son las matriculas?'
)

print(response.text)
```
